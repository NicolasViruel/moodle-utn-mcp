<div style="font-family:'Segoe UI',Arial,sans-serif; color:#1a237e; line-height:1.7;">
  <h2 style="color:#1565c0; font-size:1.35rem;">1. Preparar un archivo de ejemplo</h2>
  <p>
    Para que el notebook pueda ejecutarse sin descargar archivos adicionales, la siguiente
    celda crea un pequeño archivo llamado <code>reservas_ejemplo.csv</code>.
  </p>
  <p>
    El archivo contiene datos simulados de reservas hoteleras. Se utiliza únicamente para
    aprender el procedimiento básico de carga e inspección.
  </p>
</div>


```python
# Ejecutá esta celda para crear el archivo de ejemplo.
from io import StringIO
from pathlib import Path

contenido_csv = """reserva_id,tipo_hotel,noches,adultos,canal_reserva,precio_promedio,is_canceled
1001,Urbano,2,2,Directo,92.50,0
1002,Resort,5,2,Agencia,145.00,1
1003,Urbano,1,1,Directo,78.00,0
1004,Resort,7,2,Online,168.40,0
1005,Urbano,3,2,Online,110.00,1
1006,Resort,4,3,Agencia,152.75,0
1007,Urbano,2,1,Online,88.20,0
1008,Resort,6,2,Directo,175.00,1
1009,Urbano,1,2,Agencia,83.50,0
1010,Resort,3,2,Online,139.90,0
1011,Urbano,4,2,Directo,118.00,1
1012,Resort,5,4,Agencia,189.50,0
"""

ruta_archivo = Path("reservas_ejemplo.csv")
ruta_archivo.write_text(contenido_csv, encoding="utf-8")

print(f"Archivo creado correctamente: {ruta_archivo.name}")
```

  🔎 <strong>Qué observar:</strong> la salida confirma que el archivo fue creado en el
  entorno de trabajo. En una situación real, el CSV podría provenir de Moodle, Google Drive
  u otra fuente.
</div>



<div style="font-family:'Segoe UI',Arial,sans-serif; color:#1a237e; line-height:1.7;">
  <h2 style="color:#1565c0; font-size:1.35rem;">2. Importar Pandas</h2>
  <p>
    Pandas es una librería de Python utilizada para organizar y analizar datos.
    Habitualmente se importa utilizando el alias <code>pd</code>.
  </p>
</div>


```python
import pandas as pd

print("Versión de Pandas:", pd.__version__)
```


<div style="background-color:#e3f2fd; border:1px solid #bbdefb; border-radius:8px; padding:12px 14px; font-family:'Segoe UI',Arial,sans-serif; color:#1a237e; line-height:1.7;">
  💡 <strong>Idea clave:</strong> importar una librería permite utilizar las funciones y
  estructuras que esa herramienta ofrece. El alias <code>pd</code> simplifica la escritura
  de los comandos.
</div>



<div style="font-family:'Segoe UI',Arial,sans-serif; color:#1a237e; line-height:1.7;">
  <h2 style="color:#1565c0; font-size:1.35rem;">3. Cargar el archivo en un DataFrame</h2>
  <p>
    La función <code>pd.read_csv()</code> lee el archivo y lo convierte en un
    <strong>DataFrame</strong>: una estructura organizada en filas y columnas.
  </p>
</div>


```python
df = pd.read_csv("reservas_ejemplo.csv")

print("El dataset fue cargado correctamente.")
```


<div style="background-color:#ffffff; border:1px solid #cfd8dc; border-radius:8px; padding:12px 14px; font-family:'Segoe UI',Arial,sans-serif; color:#1a237e; line-height:1.7;">
  📌 El nombre <code>df</code> es una convención frecuente para referirse a un DataFrame.
  Podrían utilizarse otros nombres, siempre que resulten claros y descriptivos.
</div>



<div style="font-family:'Segoe UI',Arial,sans-serif; color:#1a237e; line-height:1.7;">
  <h2 style="color:#1565c0; font-size:1.35rem;">4. Observar las primeras filas</h2>
  <p>
    El método <code>head()</code> permite realizar una primera lectura de la estructura y
    del contenido del dataset. De manera predeterminada muestra las primeras cinco filas.
  </p>
</div>


```python
df.head()
```


<div style="background-color:#e3f2fd; border:1px solid #bbdefb; border-radius:8px; padding:12px 14px; font-family:'Segoe UI',Arial,sans-serif; color:#1a237e; line-height:1.7;">
  🔎 <strong>Preguntas para mirar la salida:</strong>
  <ul style="margin-bottom:0;">
    <li>¿Qué representa cada fila?</li>
    <li>¿Qué información parece contener cada columna?</li>
    <li>¿Qué valores aparecen en <code>is_canceled</code>?</li>
  </ul>
</div>



<div style="font-family:'Segoe UI',Arial,sans-serif; color:#1a237e; line-height:1.7;">
  <h2 style="color:#1565c0; font-size:1.35rem;">5. Consultar las dimensiones</h2>
  <p>
    El atributo <code>shape</code> devuelve dos valores: la cantidad de filas y la cantidad
    de columnas del DataFrame.
  </p>
</div>


```python
filas, columnas = df.shape

print("Cantidad de filas:", filas)
print("Cantidad de columnas:", columnas)
```


<div style="background-color:#ffffff; border:1px solid #cfd8dc; border-radius:8px; padding:12px 14px; font-family:'Segoe UI',Arial,sans-serif; color:#1a237e; line-height:1.7;">
  💡 <strong>Interpretación:</strong> las filas suelen representar observaciones o registros;
  las columnas representan variables o atributos.
</div>



<div style="font-family:'Segoe UI',Arial,sans-serif; color:#1a237e; line-height:1.7;">
  <h2 style="color:#1565c0; font-size:1.35rem;">6. Reconocer columnas y tipos de datos</h2>
  <p>
    <code>columns</code> muestra los nombres de las variables y <code>dtypes</code> informa
    el tipo de dato asignado por Pandas a cada columna.
  </p>
</div>


```python
print("Columnas del dataset:")
print(df.columns.tolist())

print("\nTipos de datos:")
print(df.dtypes)
```


<div style="background-color:#e3f2fd; border:1px solid #bbdefb; border-radius:8px; padding:12px 14px; font-family:'Segoe UI',Arial,sans-serif; color:#1a237e; line-height:1.7;">
  📌 <strong>Algunos tipos frecuentes:</strong>
  <ul style="margin-bottom:0;">
    <li><code>int64</code>: números enteros.</li>
    <li><code>float64</code>: números con decimales.</li>
    <li><code>object</code>: texto o categorías.</li>
  </ul>
</div>



<div style="font-family:'Segoe UI',Arial,sans-serif; color:#1a237e; line-height:1.7;">
  <h2 style="color:#1565c0; font-size:1.35rem;">7. Obtener una vista general con info()</h2>
  <p>
    El método <code>info()</code> reúne información sobre el tamaño del DataFrame, las
    columnas, la cantidad de valores no nulos y los tipos de datos.
  </p>
</div>


```python
df.info()
```


<div style="background-color:#ffffff; border:1px solid #cfd8dc; border-radius:8px; padding:12px 14px; font-family:'Segoe UI',Arial,sans-serif; color:#1a237e; line-height:1.7;">
  🔎 <strong>Qué observar:</strong> en este archivo de ejemplo todas las columnas tienen
  12 valores no nulos. Más adelante aprenderás a estudiar con mayor profundidad la calidad
  de los datos.
</div>



<div style="font-family:'Segoe UI',Arial,sans-serif; color:#1a237e; line-height:1.7;">
  <h2 style="color:#1565c0; font-size:1.35rem;">8. Obtener un resumen descriptivo</h2>
  <p>
    <code>describe()</code> calcula automáticamente algunas medidas descriptivas para las
    columnas numéricas: cantidad de datos, media, desviación estándar, mínimo, cuartiles y
    máximo.
  </p>
</div>


```python
df.describe().round(2)
```


<div style="background-color:#e3f2fd; border:1px solid #bbdefb; border-radius:8px; padding:12px 14px; font-family:'Segoe UI',Arial,sans-serif; color:#1a237e; line-height:1.7;">
  💡 <strong>Idea clave:</strong> el resultado numérico no constituye por sí solo una
  conclusión. Debemos interpretar qué describe cada medida y relacionarla con el contexto
  del problema.
</div>
