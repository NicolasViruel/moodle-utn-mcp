# 1. Definicion del problema

El problema seleccionado para el Trabajo Práctico Integrador consiste en analizar
las cancelaciones de reservas hoteleras.

Las cancelaciones representan una problemática relevante para los hoteles debido
a que pueden afectar la planificación de la ocupación, la disponibilidad de
habitaciones y los ingresos esperados.

El objetivo inicial del análisis será estudiar las características de las reservas
y buscar identificar qué factores pueden estar relacionados con la cancelación
de una reserva.

# 2. Variable objetivo

La variable objetivo del problema es **is_canceled**.

Esta variable indica si una reserva hotelera fue cancelada o no.

- `is_canceled = 0`: la reserva no fue cancelada.
- `is_canceled = 1`: la reserva fue cancelada.

Por lo tanto, el análisis buscará estudiar qué características de las reservas
pueden estar relacionadas con el valor de esta variable.


# 3. Preguntas iniciales de análisis

1. ¿Qué **porcentaje** de reservas se cancela en el dataset?
2. ¿Las cancelaciones varían según el **tipo de deposito**, **canal de reserva** o **anticipación** (lead time)?
3. ¿Hay diferencias por **temporada**, **tipo de habitación** o **país** del cliente?
4. ¿Clientes **repetidos** cancelan menos que clientes nuevos?
5. ¿Qué variables parecen más relacionadas con `is_canceled` en una primera exploración?
6. ¿Qué decisiones concretas podría tomar el hotel si confirma ciertos patrones?
